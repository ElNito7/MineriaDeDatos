```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Definimos las dimensiones del modelo (Notación ISLP)
n = 5 # Número de observaciones (Usuarios)
p = 3 # Dimensionalidad (Features: Edad, Salario_USD, Engagement_Score)

# Construimos la Matriz X (n x p) con datos de juguete
X_matriz = np.array([
    [25, 3000, 8.5],  # Usuario 1: Vector en 3D
    [45, 8000, 4.2],  # Usuario 2
    [22, 1500, 9.8],  # Usuario 3
    [35, 5000, 6.1],  # Usuario 4
    [28, 4200, 7.5]   # Usuario 5
])

# Para comodidad humana, la envolvemos en un DataFrame
df_X = pd.DataFrame(X_matriz, columns=['Edad', 'Salario', 'Engagement'], 
                    index=['U1', 'U2', 'U3', 'U4', 'U5'])

print("Nuestra Matriz Multidimensional X (n=5, p=3):")
display(df_X)
```

```python
# Extrayendo los vectores de los usuarios
vector_u2 = X_matriz[1] 
vector_u4 = X_matriz[3] 

print(f"Coordenadas Usuario 2: {vector_u2}")
print(f"Coordenadas Usuario 4: {vector_u4}")

# Calculando la distancia euclideana multidimensional pura (Norma L2)
distancia = np.linalg.norm(vector_u2 - vector_u4)
print(f"\nLa distancia geométrica exacta entre estos dos usuarios es: {distancia:.2f} unidades.")
print("*(Nota mental: ¿Notan cómo el salario está dominando el cálculo? ¡Por eso vimos el escalamiento la semana pasada!)*")
```

```python
z_matrix = (X_matriz - X_matriz.mean()) / X_matriz.std()

z_vector_u2 = z_matrix[1]
z_vector_u4 = z_matrix[3]

print(f"Coordenadas Usuario 2: {z_vector_u2}")
print(f"Coordenadas Usuario 4: {z_vector_u4}")

# Calculando la distancia euclideana multidimensional pura (Norma L2)
distancia = np.linalg.norm(z_vector_u2 - z_vector_u4)
print(f"\nLa distancia geométrica exacta entre estos dos usuarios es: {distancia:.2f} unidades.")
```

# ¿Porqué cambió la distancia? 
Cambió porque usamos la estandarización Z-Score para escalar los datos teniendo una media de 0 y una desviación estandar de 1. Con esta transformación prevenimos que los datos con valores muy grandes dominen el cálculo. Esta es la métrica que toda IA debe usar, ya que de lo contrario el modelo puede verse sesgado por los valores muy grandes, llegando a "ignorar" a los valores pequeños en sus cálculos.
