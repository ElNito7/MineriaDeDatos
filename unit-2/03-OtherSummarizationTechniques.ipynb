```python
import pandas as pd


url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)

df['Age'].skew()
```

- Asimetría (Skewness): Investiguen cómo aplicar la función .skew() en la columna Age. Si el resultado es positivo o negativo, ¿qué significa físicamente sobre la edad de los pasajeros en el barco? (¿Había más jóvenes o más ancianos?). 

La función .skew() mide la asimetría de un conjunto de datos. Esto nos dice que tan simétrica o inclinada está la distribución de datos respecto a su media. Si es 0, son simétricos. Por otra parte, si es positiva (>0) la cola del lado derecho es más larga y hay más valores concentrados hacia la izquierda. Finalmente, si es negativa (<0), la cola del lado izquierdo es más larga y hay más valores concentrados hacia la derecha. 

En el caso de la edad, hay una asimetría positiva, por lo que podemos decir que había más jovenes que ancianos a bordo. 

```python
df['Fare'].kurt()
```

- Curtosis (Kurtosis): Investiguen cómo aplicar la función .kurt() en la columna Fare. ¿Qué significa este número sobre el "grosor" de las colas de la distribución? ¿Confirma esto la existencia de los outliers que encontramos hoy en clase?

La función .kurt() calcula la curtosis, la cual mide que tan puntiaguda o plana está la forma de distribución de datos y el peso o tamaño de las colas en comparación con una distribución normal. Si el valor = 0, quiere decir que hay una distribución normal. Si es positivo, hay pico alto y colas pesadas, lo que indica mayor concentración de datos extremos. SI es negativo es distribucion plana y colas ligeras, lo que indica menor concentración de datos extremos. 

Como en el caso de la columna Fare el valor es positivo y muy alto (33.39), sabemos que hay pico extremadamente alto y una concentración de datos extremos sumamente pesada, por lo que si se confirma la existencia de outliers.
