```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/googleplaystore.csv"
df = pd.read_csv(url)

df.describe()
```

```python
print(df.isnull().sum())
```

- Pregunta A: Ejecuten df.isnull().sum(). Identificarán que la columna Rating tiene casi 1,500 valores nulos, mientras que Type y Android Ver tienen solo 1 o 2. ¿Por qué es arquitectónicamente seguro eliminar las filas donde falta Android Ver, pero sería un error fatal eliminar las filas donde falta Rating?

Esto es porque hay varios casos en los que los usuarios no dejan un rating, lo cual explica el porque son nulos, son datos MNAR. Por otra parte, las filas con Android ver como nulo son MAR por lo que no pasa nada si las eliminamos.

```python
print(df.count())
df = df.dropna(axis=0, subset=['Android Ver', 'Type'])

print(df.count())
```

- Pregunta B: Escriban la instrucción exacta usando dropna() pasándole el parámetro subset=['Android Ver', 'Type'] para eliminar únicamente las 3 filas que tienen esos nulos. Verifiquen que el total de registros en el DataFrame apenas disminuyó.

Si. El número de elementos que disminuyeron fueron pocos. 

```python
df['Rating'].fillna(df.groupby('Category')['Rating'].transform('median'), inplace=True)

print(df.isnull().sum())
```

- Pregunta D: Revisen la columna Size (Tamaño de la app). Notarán que no tiene nulos matemáticos (NaN), pero muchas filas dicen "Varies with device". Para un modelo matemático, ese texto es basura indigerible. Escriban una instrucción usando df['Size'].replace(...) para convertir exactamente esa frase de texto en un nulo matemático real usando np.nan de NumPy. ¿Por qué es mejor convertirlo a NaN en esta etapa en lugar de ponerle un tamaño falso como "0 MB"?

Es mejor convertirlo en NaN porque si lo cambiaramos por un valor falso cambiariamos la distribución de los datos, es mejor tomarlos como nulos ya que no los podemos medir de forma confiable e ingorarlos o procesarlos de otra forma, como usando la mediana. Así mantenemos una buena representación de los datos y evitamos problemas en el futuro.
