```python
import random 

random_list = [random.random() for _ in range(5_000_000)]

print("random list length:", len(random_list))
```

```python
def sd(nums):
    avg = sum(nums)/len(nums)

    summ = 0

    for i in range(len(nums)):
        val = (nums[i]-avg)**2

        summ += val

    var = summ/len(nums)

    return var**0.5

print("random list standard deviation:", sd(random_list))
```

```python
import numpy as np 

arr = np.array(random_list)

print("random list standard deviation:", arr.std())
```

```python
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)

df.groupby('Sex')['Fare'].mean()

```

```python
# Outliers
q1 = df['Fare'].quantile(0.25)
q3 = df['Fare'].quantile(0.75)
icr = q3 - q1

min_outlier = q1 - 1.5*icr
max_outlier = q3 + 1.5*icr

print(f"El outlier minimo en fare es de {min_outlier} y el maximo de {max_outlier}")
```
